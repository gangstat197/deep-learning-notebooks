```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class LinearRegression:
    def __init__(self, num_features=1, seed=42):
        rng = np.random.default_rng(seed)
        self.W = rng.uniform(0.0, 1.0, size=(1, num_features))
        self.b = float(rng.uniform(0.0, 1.0))
        self.lr = 1e-6 
        
    def forward(self, X):
        # (1, num_features) * (num_features, N) + (1) -- broadcast b 
        # Y_pred.shape = (1, N)
        self.X = X
        return self.W @ X + self.b 

    def loss(self, Y_pred, Y):
        # (1, N)
        self.e = Y - Y_pred # error

        # float  
        return float(np.sum(self.e ** 2)) # loss
        # MSE: 1/2 * (self.e ** 2) / N 
 
    def backward(self):
        # (1, N) * (N, num_features)
        self.dW = 2 * self.e @ -(self.X.T)
        # e = Y - Y_pred; Y_pred = W.X + b => de/db = -1
        # dL/dB = dL/de * de/db = 2*e.T * (-1)
        self.db = float(np.sum(-2 * self.e))

    def step(self):
        self.W -= self.lr * self.dW 
        self.b -= self.lr * self.db

    def fit(self, X, Y, epochs=1): 
        for epoch in range(epochs):
            Y_pred = self.forward(X)
            loss_value = self.loss(Y_pred, Y)
            self.backward()
            self.step()

            print(f"Epoch: {epoch}, Loss: {loss_value}")

    # WRONG IMPLEMENTATION: I perturbed X instead of W & b
    # approximation: dL/dW = [L(W + h) - L(W - h)] / 2h  
    # def gradient_check(self, Y_pred):
    #     h0 = 1e-4
    #     for f in range(self.X.shape[0]):
    #         Y_h0 = self.forward(X[f] + h0)
    #         check = (Y_pred - Y_h0) / h0 >= 1e-6

    #         if True in check:
    #             return False

    #     raise True
```
